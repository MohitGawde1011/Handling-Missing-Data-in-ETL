1.  What are the most common reasons for missing data in ETL pipelines?

- Common reasons include:

Data entry errors – Users may leave fields blank.
System failures – Data may not be captured due to technical problems.
Optional fields – Some fields are not mandatory.
Integration issues – Data from different systems may not contain the same columns.
Data corruption – Records may be lost or damaged during processing.
API/source problems – External APIs may fail to provide certain values.
Privacy restrictions – Some customer information may intentionally be unavailable.


2. Why is blindly deleting rows with missing values considered a bad practice in ETL?

-  Blindly deleting rows can cause loss of valuable information and reduce the size of the dataset. If missing values are concentrated in a particular group, deleting those rows can also introduce bias into the analysis.

For example, if many high-income customers have missing income values and these rows are deleted, the calculated average income may become inaccurate.

3. Explain the difference between Listwise deletion and Column deletion.

-  | Method                | Meaning                                                                             | Appropriate Scenario                                                                                    |
| --------------------- | ----------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------- |
| **Listwise deletion** | Removes the entire row if one or more values are missing.                           | Appropriate when only a very small number of rows have missing values and those rows are not important. |
| **Column deletion**   | Removes an entire column when it contains too many missing values or is not useful. | Appropriate when a column has extremely high missingness and provides little analytical value.          |


Example:
If 2 out of 10,000 customer records have missing ages, listwise deletion may be acceptable.
If a column such as SecondaryPhoneNumber is missing for 95% of customers and is not needed, column deletion may be appropriate.

4. Why is median imputation preferred over mean imputation for skewed data such as income?

-  Income data is usually skewed, because a small number of people may have extremely high incomes.

The mean is strongly affected by these extreme values, while the median is more resistant to outliers.

Example:
Income: ₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹5,00,000

Mean = ₹1,22,000
Median = ₹30,000

The median gives a more realistic representation of the typical customer income. Therefore, median imputation is generally preferred for highly skewed data.

5. What is forward fill and in what type of dataset is it most useful?

- Forward fill (FFILL) replaces a missing value with the most recent available value before it.

Example:
| Date      | Temperature |
| --------- | ----------: |
| Monday    |        30°C |
| Tuesday   |     Missing |
| Wednesday |        32°C |


After forward fill:
| Date      | Temperature |
| --------- | ----------: |
| Monday    |        30°C |
| Tuesday   |        30°C |
| Wednesday |        32°C |


It is most useful for time-series or sequential datasets, such as stock prices, sensor readings, transaction data, and daily measurements, where the previous value is reasonably expected to remain valid temporarily.

6. Why should flagging missing values be done before imputation in an ETL workflow?

- A missing-value flag records whether the original value was missing before it is replaced.

For example:

|  Income | Income_Missing |
| ------: | -------------: |
| ₹30,000 |              0 |
| ₹35,000 |              0 |
| ₹32,000 |              1 |


After imputation, the missing income can be replaced with ₹32,000, but the flag still shows that the original value was missing.

This is useful because:

It preserves information about the original data quality.
Analysts can identify patterns in missingness.
Missingness itself may be predictive of business outcomes.
It provides better data lineage and auditability.

Therefore, the flag should be created before imputation, otherwise the original missingness information may be lost.

7. Consider a scenario where income is missing for many customers. How can this missingness itself provide business insights?

- Missing income values can reveal important customer behavior or data-collection problems.

For example:

Customers with lower trust in the company may be less willing to provide income information.
Certain customer groups may have higher missing-income rates.
Income may be missing because a particular sales channel does not collect it.
New customers may be more likely to have incomplete profiles.
Customers with missing income may respond differently to marketing campaigns.

Therefore, instead of simply filling or deleting the missing values, businesses should analyze the pattern of missingness. It can help identify customer segments, improve data-collection processes, and support better business decisions.
